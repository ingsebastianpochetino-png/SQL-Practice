```python
import pandas as pd
import sqlite3
import os

DB_PATH = "practica_dml.db"

if os.path.exists(DB_PATH):
    os.remove(DB_PATH)
    print("Base de datos anterior eliminada")
else:
    print("No existía una base de datos anterior")

conn = sqlite3.connect("practica_dml.db")
cursor = conn.cursor()

```

```python
cursor.execute ("""
CREATE TABLE vendedores (
id INTEGER PRIMARY KEY,
nombre TEXT NOT NULL,
ventas_totales REAL NOT NULL,
zona TEXT NOT NULL,
fecha_ingreso DATE NOT NULL
)
""")

conn.commit()

print("Base de datos y tabla creadas correctamente.")
```

```python
# Insertamos algunos registros en la base

cursor.execute("""
INSERT INTO vendedores (id, nombre, ventas_totales, zona, fecha_ingreso)
VALUES
    (1, 'Juan Perez', 8500, 'Norte', '2024-01-15'),
    (2, 'Maria Gomez', 9200, 'Sur', '2023-08-10'),
    (3, 'Carlos Lopez', 12500, 'Norte', '2022-05-20'),
    (4, 'Ana Martinez', 7000, 'Centro', '2024-03-12')
""")

conn.commit()
```

Imagina que el vendedor con id = 2 ha realizado una nueva venta. Actualiza su campo ventas_totales sumándole 500 unidades.
Reto: Cambia la zona a 'Internacional' para todos los vendedores que tengan más de 10,000 en ventas totales.

```python
# UPDATE de valores del id=2
#revisamos cuanto tiene hoy

ventas_antes = pd.read_sql_query("""
SELECT * FROM vendedores
WHERE id = 2
""",conn)

ventas_antes

cursor.execute("""
UPDATE vendedores
SET ventas_totales = ventas_totales + 500
WHERE id = 2
""")

conn.commit()


#revisamos cuanto le quedo despues
ventas_despues = pd.read_sql_query("""
SELECT * FROM vendedores
WHERE id = 2
""",conn)

display(ventas_antes)
display(ventas_despues)
```

Cambiar la zona a 'Internacional' para todos los vendedores que tengan más de 10.000 en ventas totales.

```python
cursor.execute("""
UPDATE vendedores
SET zona = 'Internacional'
WHERE ventas_totales > 10000
""")
conn.commit()

df = pd.read_sql_query("""
    SELECT id, nombre, ventas_totales, zona
    FROM vendedores
""", conn)

df
```
