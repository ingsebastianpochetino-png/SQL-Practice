Ejercicio SQL — Ventas de tecnología

Exploración y manipulación de datos utilizando SQLite.

```python
import sqlite3

conexion = sqlite3.connect(":memory:")

cursor = conexion.cursor()

cursor.execute("""
CREATE TABLE ventas_tecnologia (
    id_venta INTEGER,
    producto TEXT,
    categoria TEXT,
    precio_unitario REAL,
    cantidad INTEGER,
    fecha TEXT,
    pais TEXT
)
""")



datos = [
    (1, "Notebook", "Computación", 1200, 2, "2026-01-10", "Argentina"),
    (2, "Mouse", "Accesorios", 50, 5, "2026-01-11", "Colombia"),
    (3, "Monitor", "Computación", 600, 3, "2026-01-12", "Colombia"),
    (4, "Teclado", "Accesorios", 80, 4, "2026-01-13", "Argentina"),
    (5, "Celular", "Telefonía", 900, 2, "2026-01-14", "Colombia"),
    (6, "Tablet", "Telefonía", 700, 1, "2026-01-15", "Argentina"),
    (7, "Auriculares", None, 150, 3, "2026-01-16", "Colombia"),
    (8, "Smartwatch", "Wearables", 300, 2, "2026-01-17", "Argentina"),
    (9, "Cámara", "Fotografía", 1500, 1, "2026-01-18", "Colombia"),
    (10, "Parlante", "Audio", 250, 4, "2026-01-19", "Argentina")
]

cursor.executemany("""
INSERT INTO ventas_tecnologia
VALUES (?, ?, ?, ?, ?, ?, ?)
""", datos)

conexion.commit()

```

¿Qué productos tenemos y cuál es su precio, ordenados alfabéticamente por nombre?


```python
cursor.execute(
    "SELECT producto, precio_unitario from ventas_tecnologia order by producto ASC;")

resultado = cursor.fetchall()
print(resultado)
```

¿Cuáles fueron las ventas realizadas en Colombia donde el precio unitario fue mayor a 500?

```python
cursor.execute("""
SELECT *
FROM ventas_tecnologia
WHERE pais = 'Colombia'
AND precio_unitario > 500
""")

resultado = cursor.fetchall()
print(resultado)
```

Búsqueda de valores nulos

```python
cursor.execute("""
SELECT *
FROM ventas_tecnologia
WHERE categoria IS NULL
""")
resultado = cursor.fetchall()
print(resultado)
```

¿Cuáles son los ingresos totales por categoría?

```python
cursor.execute("""
SELECT categoria, sum(precio_unitario*cantidad) as ingresos_totales
FROM ventas_tecnologia
group by categoria
""")

resultado = cursor.fetchall()
print(resultado)
```

¿Qué categorías generaron más de $10.000 de ingresos?

```python
cursor.execute("""
SELECT categoria, sum(precio_unitario*cantidad) as ingresos_totales
FROM ventas_tecnologia
group by categoria
HAVING ingresos_totales > 10000
""")

resultado = cursor.fetchall()
print(resultado)
```
